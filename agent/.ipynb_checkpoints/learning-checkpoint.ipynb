```python
import os
os.environ["PYDANTIC_AI_NO_BANNER"] = "1"

from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.ollama import OllamaProvider

# Local Ollama server (no API key needed). Swap "mistral" for any model in `ollama list`.
# model = OpenAIChatModel(
#     "mistral",
#     provider=OllamaProvider(base_url="http://localhost:11434/v1"),
# )

# agent = Agent(model, instructions="Be concise.")

agent = Agent(
    "mistral",
    system_prompt='You are a helpful assistant, Always answer with a single concise sentence.',
)



# Jupyter already runs an event loop, so use `await agent.run(...)` here instead of run_sync
result = await agent.run("Say hello in 5 words.")
print(result.output)
```
